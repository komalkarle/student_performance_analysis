```python
avg_score=df.groupby("study_hours")["exam_score"].mean()
avg_score.plot(marker="o")

plt.title("Avg Exam Score vs Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Avg Exam Score")
plt.grid(True)
```

```python
plt.figure(figsize=(8,5))

sns.regplot(
    x="study_hours",
    y="exam_score",
    data=df,
    scatter_kws={"alpha":0.3},
    line_kws={"color":"red"}
)

plt.title("Study Hours vs Exam Score")
plt.show()
```
